# DATA CLEANING

## Xóa cột chứa quá nhiều missing value và xóa hàng missing value của description

```python
df.drop(columns="CustomerID", inplace=True)
df.dropna(subset='Description', inplace= True)
```

```python
print(df.isnull().sum())
```

## Xóa các hàng có quanity và unitprice không thực tế

```python
col_drop1 = df[df['Quantity'] <= 0].index
col_drop2 = df[df['UnitPrice'] <= 0].index

df.drop(col_drop1, inplace=True)
df.drop(col_drop2, inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

## Chuẩn hóa cột description (xóa khoảng trắng và viết hoa chữ đầu)

```python
df['Description'] = df['Description'].str.strip().str.title()
```

## Chuyển đổi kiểu dữ liệu sang  datatime

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], format='%Y-%m-%d')
```

```python
print(df['InvoiceDate'].dt.year)
print(df['InvoiceDate'].dt.month)
print(df['InvoiceDate'].dt.day)
```

```python
df['Revenue'] = df['Quantity'] *df['UnitPrice']
```

```python
print(df.head())
```

```python
condition = [
    df['Quantity'] <= 5,
    df['Quantity'] <= 20
]

choice = [
    "Small",
    'Medium'
]

df['order_size'] = np.select(condition, choice, default= 'Large')
```

```python
print(df.head())
```

```python
numeric_col = df.select_dtypes(['int64', 'float64']).columns

for col in numeric_col:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    
    outliers = (df[col] < lower_bound) | (df[col] > upper_bound)
    
    print(f"Biên trên của {col}: {upper_bound}")
    print(f"Biên dưới của {col}: {lower_bound}")
    
    if outliers.sum() > 0:
        print(f"Ngoại lệ xuất hiện tại cột: {col}: {outliers.sum()}")
        
    median = df[col].median()

    df[col] = df[col].apply(
        lambda x: median if (x < lower_bound or x > upper_bound) else x
    )
```
